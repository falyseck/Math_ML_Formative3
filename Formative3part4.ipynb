```python
import numpy as np
import matplotlib.pyplot as plt
import scipy

# Data
X = np.array([1.0, 3.0])
Y = np.array([3.0, 6.0])
n = len(X)

# hyperparams
alpha = 0.1
iterations = 3

# init
m = -1.0
b = 1.0
history = {'m':[], 'b':[], 'mse':[]}

def mse(m, b, X, Y):
    preds = m*X + b
    return np.mean((Y - preds)**2)

for it in range(iterations):
    preds = m*X + b
    errors = Y - preds
    # gradients
    grad_m = -2/n * np.sum(X * errors)
    grad_b = -2/n * np.sum(errors)
    # update
    m = m - alpha * grad_m
    b = b - alpha * grad_b
    # record
    history['m'].append(m)
    history['b'].append(b)
    history['mse'].append(mse(m,b,X,Y))

# final predictions
final_preds = m*X + b
print("final m, b:", m, b)
print("final preds:", final_preds)
print("final mse:", history['mse'][-1])

# plots
plt.figure(figsize=(10,4))
plt.subplot(1,2,1)
plt.plot(history['m'], label='m')
plt.plot(history['b'], label='b')
plt.xlabel('Iteration')
plt.title('m and b over iterations')
plt.legend()

plt.subplot(1,2,2)
plt.plot(history['mse'])
plt.xlabel('Iteration')
plt.title('MSE over iterations')
plt.tight_layout()
plt.show()
```
